```python
# Create a DataFrame

import pandas as pd

data = {
    "Name": ["Rahul", "Priya", "Aman"],
    "Age": [25, 28, 31],
    "Salary": [45000, 55000, 62000]
}

df = pd.DataFrame(data)

print(df)
```

```python
high_salary = df[df["Salary"] > 50000]

print(high_salary)
```

```python
df["Bonus"] = df["Salary"] * 0.10

print(df)
```
