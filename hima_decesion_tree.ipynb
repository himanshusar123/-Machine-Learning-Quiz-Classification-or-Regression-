```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.metrics import accuracy_score, confusion_matrix
from google.colab import files


# STEP 2 — Load dataset
# Replace the filename below with your uploaded file name if needed
data = pd.read_csv("customer (3).csv")

# STEP 3 — Display dataset
print("Dataset:\n")
print(data)

# STEP 4 — Scatter plot with color mapping
plt.figure(figsize=(8, 6))

purchase = data['Purchase']   # contains Yes/No
color_map = {'Yes': 'green', 'No': 'red'}
colors = purchase.map(color_map)

plt.scatter(
    data['Discount'],
    data['Income'],
    c=colors,
    s=100,
    alpha=0.8
)

plt.xlabel('Discount (%)')
plt.ylabel('Income (₹ in thousands)')
plt.title('Scatter: Discount vs Income (colored by Purchase)')
plt.grid(True)
plt.show()

# STEP 5 — Features and target
X = data[['Discount', 'Income']]
y = data['Purchase']

# STEP 6 — Train Decision Tree Model
clf = DecisionTreeClassifier(random_state=42)
clf.fit(X, y)

# STEP 7 — Prediction
predictions = clf.predict([[18, 33]])

print("\nTarget Purchase Prediction:")
print(predictions)

# STEP 8 — Evaluation
y_pred = clf.predict(X)

print("\nAccuracy Score:")
print(accuracy_score(y, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y, y_pred))

# STEP 9 — Visualize Decision Tree
plt.figure(figsize=(10, 7))

plot_tree(
    clf,
    feature_names=['Discount', 'Income'],
    class_names=['No', 'Yes'],
    filled=True,
    rounded=True
)

plt.title("Decision Tree")
plt.show()

# STEP 10 — Plot Decision Boundary
fig, ax = plt.subplots(figsize=(8, 6))

DecisionBoundaryDisplay.from_estimator(
    clf,
    X,
    cmap=plt.cm.RdYlBu,
    response_method="predict",
    ax=ax,
    xlabel="Discount (%)",
    ylabel="Income (₹ in thousands)"
)

# Original data points
plt.scatter(
    data['Discount'],
    data['Income'],
    c=colors,
    edgecolor='black',
    s=100
)

plt.title("Decision Boundary")
plt.show()
```
