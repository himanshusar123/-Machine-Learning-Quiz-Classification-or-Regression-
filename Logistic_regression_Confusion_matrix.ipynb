```python
# Telecom Customer Churn Prediction Using Logistic Regression
# Project Type

# Classification Problem (Logistic Regression)


# Business Scenario

# A telecom company is losing customers every month.

# Management wants to answer:

# Which customers are likely to leave the company?

# If the company can identify these customers in advance, it can:

# Offer discounts
# Improve customer service
# Increase retention


import pandas as pd

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)
```

```python
df = pd.read_csv(
    "Telecom_Customer_Churn_Big_Dataset.csv"
)
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df["Churn"].value_counts()
```

```python
df["Churn"] = df["Churn"].map({
    "No":0,
    "Yes":1
})
```

```python
X = df[
    [
        "Age",
        "Monthly_Charges",
        "Tenure_Months",
        "Support_Calls"
    ]
]
```

```python
y = df["Churn"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
model = LogisticRegression()

model.fit(
    X_train,
    y_train
)
```

```python
y_pred = model.predict(
    X_test
)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

print(
    "Accuracy:",
    round(accuracy*100,2),
    "%"
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)
```

```python
precision = precision_score(
    y_test,
    y_pred
)

print(
    "Precision:",
    round(precision,2)
)
```

```python
recall = recall_score(
    y_test,
    y_pred
)

print(
    "Recall:",
    round(recall,2)
)
```

```python
f1 = f1_score(
    y_test,
    y_pred
)

print(
    "F1 Score:",
    round(f1,2)
)
```

```python
print(
    classification_report(
        y_test,
        y_pred
    )
)
```
