```python
import pandas as pd

df = pd.read_csv(
    "Customer_Segmentation_Big_Dataset.csv"
)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
X = df[
    [
        "Annual_Income",
        "Spending_Score"
    ]
]
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters=5,
    random_state=42
)

df["Cluster"] = kmeans.fit_predict(X)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

plt.scatter(
    df["Annual_Income"],
    df["Spending_Score"],
    c=df["Cluster"]
)

plt.xlabel("Annual Income")

plt.ylabel("Spending Score")

plt.title("Customer Segmentation")

plt.show()
```

```python
df.groupby("Cluster")[[
    "Annual_Income",
    "Spending_Score"
]].mean()
```
