## Section 14: Adding Calculator Methods to the Class
Demonstrate adding calculations directly inside the class as instance methods.


```python
class Employee:

    def annual_salary(self):
        return self.salary * 12

    def calculate_bonus(self):
        return self.annual_salary() * 0.10

    def calculate_tax(self):
        return self.annual_salary() * 0.08

    def net_salary(self):
        return self.annual_salary() + self.calculate_bonus() - self.calculate_tax()
```

## Section 17: Defining Search Functionality
Define a search function that scans the list of employee objects for a specific ID.


```python
# Helper function to find employee object in the list
def search_employee(employee_id):
    for employee in employees:
        if employee.employee_id == employee_id:
            return employee
    return None
```

## Section 20: OOP and Input Integration
Redefine class and add dynamic input collection to construct and display new employee instances.


```python
employees = []

class Employee:
    def __init__(self, employee_id, name, department, designation, salary):
        self.employee_id = employee_id
        self.name = name
        self.department = department
        self.designation = designation
        self.salary = salary

    def display(self):
        print("-----------------------------")
        print("Employee ID :", self.employee_id)
        print("Name :", self.name)
        print("Department :", self.department)
        print("Designation :", self.designation)
        print("Salary :", self.salary)

# Collect fields interactively from standard input
employee_id = int(input("Employee ID : "))
name = input("Name : ")
department = input("Department : ")
designation = input("Designation : ")
salary = float(input("Salary : "))

employee = Employee(employee_id, name, department, designation, salary)
employees.append(employee)

print("\nEmployee Added Successfully!")

for emp in employees:
    emp.display()
```

## Section 22: Final Integration - Menu-Driven HR System
A fully functional console-based menu system integrating employee creation, viewing, lookup, financial reporting, and server integrations.


```python
# Final Integration - Complete Menu-Driven HR System

while True:
    print()
    print("RetailMax HR Automation")
    print("1 Add Employee")
    print("2 Display Employees")
    print("3 Search Employee")
    print("4 Salary Report")
    print("5 Load API Employees")
    print("6 Exit")

    choice = input("Choice : ")

    if choice == "1":
        # Add Employee with input validation
        while True:
            try:
                emp_id = int(input("Enter Employee ID : "))
                if any(emp.employee_id == emp_id for emp in employees):
                    print("Error: Employee ID already exists. Enter a unique ID.")
                    continue
                break
            except ValueError:
                print("Please enter a valid numeric Employee ID.")

        name = input("Enter Name : ")
        department = input("Enter Department : ")
        designation = input("Enter Designation : ")

        while True:
            try:
                salary = float(input("Enter Salary : "))
                if salary <= 0:
                    raise ValueError
                break
            except ValueError:
                print("Please Enter Valid Salary (positive number).")

        emp = Employee(emp_id, name, department, designation, salary)
        employees.append(emp)
        print("\nEmployee Added Successfully!")

    elif choice == "2":
        # Display All Employees
        if not employees:
            print("\nNo employees available in database.")
        else:
            for emp in employees:
                emp.display()

    elif choice == "3":
        # Search Employee by ID
        try:
            search_id = int(input("Enter Employee ID to search : "))
            emp = search_employee(search_id)
            if emp:
                emp.display()
            else:
                print("Employee Not Found")
        except ValueError:
            print("Please enter a valid numeric ID.")

    elif choice == "4":
        # Print Annual Salary Reports for all employees
        if not employees:
            print("\nNo employees available to report.")
        else:
            for emp in employees:
                annual = emp.salary * 12
                bonus = annual * 0.10
                tax = annual * 0.08
                net = annual + bonus - tax
                print("-"*30)
                print(f"Employee : {emp.name} (ID: {emp.employee_id})")
                print(f"Annual Salary : {annual}")
                print(f"Bonus : {bonus}")
                print(f"Tax : {tax}")
                print(f"Net : {net}")

    elif choice == "5":
        # Load API Employees and display them
        print("\nLoading employee records from REST API...")
        try:
            import requests
            url = "https://jsonplaceholder.typicode.com/users"
            response = requests.get(url)
            users = response.json()
            print("\nFetched names from server API:")
            print("-"*30)
            for user in users:
                print(user["name"])
                # Mock add to list to demonstrate integration
                mock_id = 2000 + user["id"]
                if not any(emp.employee_id == mock_id for emp in employees):
                    company = user.get("company", {}).get("name", "External API")
                    role = user.get("company", {}).get("bs", "Consultant")
                    mock_salary = 50000.0 + user["id"] * 2500
                    mock_emp = Employee(mock_id, user["name"], company, role, mock_salary)
                    employees.append(mock_emp)
            print("\nAPI Employees integrated successfully!")
        except Exception as e:
            print("Error loading API data:", e)

    elif choice == "6":
        print("Exiting RetailMax HR Automation System. Goodbye!")
        break
```
