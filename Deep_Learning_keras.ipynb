```python
# STEP 1 — Import Required Libraries

import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.linear_model import LinearRegression


# STEP 2 — Upload CSV File

uploaded = files.upload()


# STEP 3 — Load Dataset

df = pd.read_csv("student_marks (3).csv")


# STEP 4 — Display Dataset

print("Dataset:\n")

print(df)


# STEP 5 — Check Column Names

print("\nColumn Names:\n")

print(df.columns)


# STEP 6 — Visualize Data

plt.scatter(df['StudyHours'], df['Marks'])

plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.title("Study Hours vs Marks")

plt.show()


# STEP 7 — Prepare Input and Output Data

X = df[['StudyHours']]   # Input Feature
y = df['Marks']          # Output Feature


# STEP 8 — Create Linear Regression Model

model = LinearRegression()


# STEP 9 — Train the Model

model.fit(X, y)


# STEP 10 — Predict Marks for 7 Study Hours

prediction = model.predict([[7]])

print("\nPredicted Marks for 7 Study Hours:\n")

print(prediction)


# STEP 11 — Draw Regression Line

plt.scatter(df['StudyHours'], df['Marks'])

plt.plot(df['StudyHours'], model.predict(X))

plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.title("Linear Regression Line")

plt.show()


# STEP 12 — Display Model Equation Values

print("\nSlope (m):")

print(model.coef_)

print("\nIntercept (c):")

print(model.intercept_)
```
