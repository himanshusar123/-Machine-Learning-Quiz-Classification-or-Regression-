```python
# STEP 1 — Import Libraries

import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

from google.colab import files
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense


# STEP 2 — Upload Dataset

uploaded = files.upload()


# STEP 3 — Load Dataset

df = pd.read_csv("student_marks (3).csv")

print("Dataset:\n")

print(df)


# STEP 4 — Visualize Dataset

plt.scatter(df['StudyHours'], df['Marks'])

plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.title("Study Hours vs Marks")

plt.show()


# STEP 5 — Prepare Input and Output

X = df[['StudyHours']]
y = df['Marks']


# STEP 6 — Create Neural Network

model = Sequential()

# Hidden Layer
model.add(Dense(8, activation='relu', input_shape=(1,)))

# Output Layer
model.add(Dense(1))


# STEP 7 — Compile Model

model.compile(
    optimizer='adam',
    loss='mean_squared_error'
)


# STEP 8 — Train Model

model.fit(X, y, epochs=100)


# STEP 9 — Predict Marks

prediction = model.predict(np.array([[7]]))

print("\nPredicted Marks for 7 Study Hours:\n")

print(prediction)


# STEP 10 — Plot Predictions

plt.scatter(df['StudyHours'], df['Marks'])

plt.plot(df['StudyHours'], model.predict(X))

plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.title("Deep Learning Prediction")

plt.show()
```
