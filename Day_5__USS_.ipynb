```python
import matplotlib.pyplot as plt
```

```python
#LineChart

months = ["Jan","Feb","Mar","Apr","May"]
sales = [25000,28000,35000,42000,47000]

plt.plot(months, sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
#Barchart

# Department-wise Employees

departments = ["HR","IT","Finance","Sales"]
employees = [12,30,18,20]

plt.bar(departments, employees)
plt.title("Employees by Department")
plt.show()
```

```python
import matplotlib.pyplot as plt

#Piechart
# Scenario:

# Market Share

brands=["Apple","Samsung","OnePlus","Others"]
share=[40,30,20,10]

# Example: Define 'colors' variable with a list of colors
colors = ['gold', 'lightcoral', 'lightskyblue', 'lightgreen']

plt.pie(share,
labels=brands,
colors=colors, # Added a comma here and used the defined 'colors' variable
autopct="%1.1f%%")

plt.show()
```

```python
# Historam
# Scenario:

# Employee Salaries

salary=[35000,42000,50000,45000,62000,
58000,47000,39000,51000,65000]

plt.hist(salary,bins=5)

plt.show()
```

```python
# Customizing Charts
plt.plot(months,sales,
marker='o')

plt.title("Monthly Sales")

plt.xlabel("Months")

plt.ylabel("Sales")

plt.grid(True)

plt.show()
```
