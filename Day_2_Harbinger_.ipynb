# RetailMax HR Automation System - Phase 2
This notebook demonstrates file management, reading/writing CSV data, searching records, and modifying existing employee files in Python.


## Section 1: Initializing and Verifying Environment Setup
Here, we import the `os` library and verify our current working directory to ensure files are read/written correctly.


```python
import os

print("=" * 60)
print("RetailMax HR Automation System")
print("=" * 60)

# Get current working directory
current_path = os.getcwd()

print(f"Current Working Directory : {current_path}")
```

## Section 2: Checking Employee Database Existence
We verify if the `employees.csv` file exists in the directory before performing any database operations.


```python
filename = "employees.csv"

# Check if employee database exists in working directory
if os.path.exists(filename):
    print("Employee database found.")
else:
    print("Employee database NOT found.")
```

## Section 3: Retrieving Absolute Path of Database
Retrieve the full absolute file system path of the database to trace file locations.


```python
# Get absolute path of the database file
full_path = os.path.abspath(filename)

print(f"\nEmployee Database Path")
print(full_path)
```

## Section 4: Counting Existing Employees in Database
Read the CSV file as a raw list of rows and count the total number of employee records.


```python
import csv

# Open and parse database to count employees
with open(filename, "r") as file:
    reader = csv.reader(file)
    rows = list(reader)

# Subtract 1 from row length assuming a header row might be present
employee_count = len(rows) - 1

print(f"\nExisting Employees : {employee_count}")
```

## Section 5: Displaying All Employees (Raw CSV Rows)
Read the employee database and display each row in its raw list format.


```python
# Display All Employees
import csv

print("=" * 70)
print("RetailMax HR Employee Database")
print("=" * 70)

# Open and print each raw row from CSV
with open(filename, "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

## Section 6: Formatted Tabular Display of Employees
Print a formatted employee directory with proper headers and left/right aligned columns.


```python
import csv

# Display employee database in clean tabular format
with open(filename, "r") as file:
    reader = csv.reader(file)

    print("-"*75)
    print(f"{'ID':<10}{'Name':<25}{'Department':<20}{'Salary':>10}")
    print("-"*75)

    for row in reader:
        print(f"{row[0]:<10}{row[1]:<25}{row[2]:<20}{row[3]:>10}")
```

## Section 7: Searching Employee by ID
Prompts the user for a search ID and prints the corresponding employee details if found.


```python
# Searching employee by employee_id
search_id = input("Enter Employee ID : ")

found = False

# Search loop through CSV file
with open(filename, "r") as file:
    reader = csv.reader(file)
    for row in reader:
        if row[0] == search_id:
            print("\nEmployee Found")
            print("-"*40)
            print("ID :", row[0])
            print("Name :", row[1])
            print("Department :", row[2])
            print("Salary :", row[3])
            found = True
            break

if not found:
    print("Employee Not Found")
```

## Section 8: Searching Employees by Department
Prompts the user for a department name and lists all employees belonging to that department (case-insensitive).


```python
# Search employees by Department

department = input("Enter Department : ")

with open(filename,"r") as file:
    reader = csv.reader(file)
    print()
    print("Employees in", department)
    print("-"*50)
    for row in reader:
        # Check match with case-insensitive check
        if row[2].lower() == department.lower():
            print(row)
```

## Section 9: Manager Report - Average Salary Calculation
Calculates and prints the average salary across all employees stored in the database.


```python
# Manager Needs report

import csv

total_salary = 0
employee_count = 0

# Read all salaries and calculate their average
with open(filename,"r") as file:
    reader = csv.reader(file)
    for row in reader:
        total_salary += int(row[3])
        employee_count += 1

print("Average Salary :", total_salary/employee_count)
```

## Section 10: Deleting an Employee Record
Allows deleting an employee from the system by entering their Employee ID. The script filters out the matching row and writes the remaining records back to `employees.csv`.


```python
# Deleting employee
import csv
import os

filename = "employees.csv"
delete_id = input("Enter Employee ID to delete : ")

if not os.path.exists(filename):
    print("Employee database NOT found.")
else:
    updated_rows = []
    found = False

    # Read and filter out the employee to delete
    with open(filename, "r", newline="", encoding="utf-8") as file:
        reader = csv.reader(file)
        for row in reader:
            if row and row[0] == delete_id:
                found = True
                print(f"\nDeleting Employee: {row[1]} (ID: {row[0]}, Department: {row[2]}) ")
            else:
                updated_rows.append(row)

    # If employee was found, save changes back to CSV
    if found:
        with open(filename, "w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerows(updated_rows)
        print("Employee deleted successfully.")
    else:
        print("Employee Not Found")
```

## Section 11: Transferring Employee Department
Allows transferring an employee to another department by entering their Employee ID and the new department. The script updates the department column in the database and writes it back.


```python
# Transferring Employee
import csv
import os

filename = "employees.csv"
transfer_id = input("Enter Employee ID to transfer : ")
new_department = input("Enter New Department : ")

if not os.path.exists(filename):
    print("Employee database NOT found.")
else:
    updated_rows = []
    found = False

    # Read and update the department column
    with open(filename, "r", newline="", encoding="utf-8") as file:
        reader = csv.reader(file)
        for row in reader:
            if row and row[0] == transfer_id:
                found = True
                old_dept = row[2]
                row[2] = new_department
                print(f"\nTransferring Employee: {row[1]} (ID: {row[0]})")
                print(f"Old Department: {old_dept} -> New Department: {new_department}")
            updated_rows.append(row)

    # If employee was found, save updates back to CSV
    if found:
        with open(filename, "w", newline="", encoding="utf-8") as file:
            writer = csv.writer(file)
            writer.writerows(updated_rows)
        print("Employee transferred successfully.")
    else:
        print("Employee Not Found")
```
