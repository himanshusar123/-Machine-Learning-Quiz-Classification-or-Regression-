```python
# IMPORT LIBRARIES
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np
```

```python
mnist = tf.keras.datasets.mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
plt.imshow(X_train[0], cmap='gray')

plt.title("Sample Digit")

plt.axis("off")

plt.show()
```

```python
# NORMALIZE IMAGES

X_train = X_train / 255.0

X_test = X_test / 255.0
```

```python
#  NEURAL NETWORK CODE
model = tf.keras.models.Sequential()
```

```python
# ADD LAYERS CODE
model.add(
    tf.keras.layers.Flatten(input_shape=(28,28))
)

model.add(
    tf.keras.layers.Dense(
        128,
        activation='relu'
    )
)

model.add(
    tf.keras.layers.Dense(
        10,
        activation='softmax'
    )
)
```

```python
# COMPILE MODEL CODE
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
# TRAIN MODEL

model.fit(
    X_train,
    y_train,
    epochs=5
)
```

```python
# PREDICT DIGIT CODE
prediction = model.predict(X_test)

predicted_digit = np.argmax(
    prediction[0]
)
```

```python
# DISPLAY PREDICTION CODE
plt.imshow(X_test[0], cmap='gray')

plt.title(
    f"Predicted Digit: {predicted_digit}"
)

plt.axis("off")

plt.show()
```

```python
index = 5

prediction = model.predict(X_test)

predicted_digit = np.argmax(
    prediction[index]
)

plt.imshow(X_test[index], cmap='gray')

plt.title(
    f"Predicted Digit: {predicted_digit}"
)

plt.axis("off")

plt.show()
```
