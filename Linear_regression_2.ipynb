```python
import pandas as pd

data = {
    'Area': [500, 700, 900, 1100, 1300],
    'Price': [10, 14, 18, 22, 26]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.linear_model import LinearRegression
X = df[['Area']]
y = df['Price']
model = LinearRegression()

model.fit(X, y)
prediction = model.predict([[1000]])

print(prediction)
```
