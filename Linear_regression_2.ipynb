```python
import pandas as pd

data = {
    'Area': [500, 700, 900, 1100, 1300],
    'Price': [10, 14, 18, 22, 26]
}

df = pd.DataFrame(data)

print(df)
```

```python
from sklearn.linear_model import LinearRegression
X = df[['Area']]
y = df['Price']
model = LinearRegression()

model.fit(X, y)
prediction = model.predict([[1000]])

print(prediction)
```

```python
# Scenario: Predicting Loan Approval Based on Age
# Context
# A bank wants to decide whether to approve a loan application based on the
# applicant’s age. They believe there is a rule-based pattern (not strictly
# linear), so they use a Decision Tree Classifier to model the decision.

from sklearn.tree import DecisionTreeClassifier

X = [[25], [35], [45], [20],[65],[70]]
y = ['No', 'Yes', 'Yes', 'No','No','No']

model = DecisionTreeClassifier()

model.fit(X, y)

print(model.predict([[67]]))
```

```python
# Scenario: Customer Segmentation Based on Shopping Behavior
# Context
# A retail store wants to group its customers into segments based on their shopping patterns.
#  They collected data on two features for each customer:

# Number of items purchased

# Total amount spent (in ₹ thousands)

# They suspect there are two distinct customer groups:

# Budget shoppers (buy fewer items, spend less)

# Premium shoppers (buy more items, spend more)

# To identify these groups, they use KMeans clustering.


from sklearn.cluster import KMeans
import numpy as np

# Step 1: Dataset
data = np.array([
    [1, 2],   # Few items, low spend
    [2, 3],   # Few items, low spend
    [8, 8],   # Many items, high spend
    [9, 10]   # Many items, high spend
])

# Step 2: Train KMeans with 2 clusters
model = KMeans(n_clusters=2)
model.fit(data)

# Step 3: Output cluster labels
print("Customer Segments:", model.labels_)
```

```python
# STEP 1 — Import required libraries

import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.cluster import KMeans


# STEP 2 — Upload CSV file

# uploaded = files.upload()


# STEP 3 — Load dataset

df = pd.read_csv("customer_data.csv")


# STEP 4 — Display dataset

print("Dataset:\n")
print(df)


# STEP 5 — Select required columns

X = df[['AnnualIncome', 'SpendingScore']]


# STEP 6 — Visualize the original data

plt.scatter(X['AnnualIncome'], X['SpendingScore'])

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")

plt.title("Customer Data")

plt.show()


# STEP 7 — Create K-Means model

model = KMeans(n_clusters=3)


# STEP 8 — Train the model

model.fit(X)


# STEP 9 — Get cluster labels

clusters = model.labels_

print("\nCluster Labels:\n")
print(clusters)


# STEP 10 — Add cluster labels to dataset

df['Cluster'] = clusters

print("\nUpdated Dataset:\n")
print(df)


# STEP 11 — Visualize clustered data

plt.scatter(df['AnnualIncome'], df['SpendingScore'], c=df['Cluster'])

plt.xlabel("Annual Income")
plt.ylabel("Spending Score")

plt.title("K-Means Clustering")

plt.show()
```
